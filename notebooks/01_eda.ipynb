# Análisis Exploratorio de Datos (EDA) - IBM HR Analytics Employee Attrition & Performance

En este notebook, realizamos un análisis exploratorio de datos para entender los factores que influyen en la rotación de empleados.

## 1. Configuración Inicial

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Configurar estilo de gráficos
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)

# Cargar datos limpios
processed_data_path = os.path.join("data", "processed", "clean_data.csv")
df = pd.read_csv(processed_data_path)

# Mostrar primeras filas
df.head()
```

## 2. Análisis de la Variable Objetivo (Attrition)

```python
# Distribución de la variable objetivo
attrition_counts = df["Attrition"].value_counts()
attrition_percent = df["Attrition"].value_counts(normalize=True) * 100

print("Distribución de Attrition:")
print(f"No (permanece): {attrition_counts[0]} ({attrition_percent[0]:.1f}%)")
print(f"Yes (se va): {attrition_counts[1]} ({attrition_percent[1]:.1f}%)")

# Gráfico de barras
plt.figure(figsize=(8, 5))
sns.countplot(x="Attrition", data=df, palette="Set2")
plt.title("Distribución de Rotación de Empleados (Attrition)")
plt.xlabel("Attrition (0 = No, 1 = Yes)")
plt.ylabel("Número de Empleados")
plt.savefig("reports/attrition_distribution.png", dpi=300, bbox_inches="tight")
plt.show()
```

## 3. Análisis de Variables Categóricas vs Attrition

```python
# Función para graficar variables categóricas vs Attrition
def plot_categorical_vs_attrition(df, col, title):
    plt.figure(figsize=(12, 6))
    ax = sns.countplot(x=col, hue="Attrition", data=df, palette="Set2")
    plt.title(title)
    plt.xlabel(col)
    plt.ylabel("Número de Empleados")
    plt.legend(title="Attrition", labels=["No", "Yes"])
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.savefig(f"reports/{col}_vs_attrition.png", dpi=300, bbox_inches="tight")
    plt.show()

# Graficar algunas variables clave
plot_categorical_vs_attrition(df, "Department", "Rotación por Departamento")
plot_categorical_vs_attrition(df, "JobRole", "Rotación por Rol de Trabajo")
plot_categorical_vs_attrition(df, "OverTime", "Rotación por Horas Extras")
plot_categorical_vs_attrition(df, "MaritalStatus", "Rotación por Estado Civil")
```

## 4. Análisis de Variables Numéricas vs Attrition

```python
# Función para graficar variables numéricas vs Attrition
def plot_numerical_vs_attrition(df, col, title):
    plt.figure(figsize=(12, 6))
    sns.boxplot(x="Attrition", y=col, data=df, palette="Set2")
    plt.title(title)
    plt.xlabel("Attrition (0 = No, 1 = Yes)")
    plt.ylabel(col)
    plt.tight_layout()
    plt.savefig(f"reports/{col}_vs_attrition.png", dpi=300, bbox_inches="tight")
    plt.show()

# Graficar algunas variables clave
plot_numerical_vs_attrition(df, "Age", "Distribución de Edad por Rotación")
plot_numerical_vs_attrition(df, "MonthlyIncome", "Ingreso Mensual por Rotación")
plot_numerical_vs_attrition(df, "DistanceFromHome", "Distancia del Hogar por Rotación")
plot_numerical_vs_attrition(df, "TotalWorkingYears", "Años Totales de Trabajo por Rotación")
```

## 5. Análisis de Correlación

```python
# Calcular matriz de correlación (solo variables numéricas)
correlation_matrix = df.select_dtypes(include=['int64', 'float64']).corr()

# Graficar heatmap
plt.figure(figsize=(16, 12))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Matriz de Correlación de Variables Numéricas")
plt.tight_layout()
plt.savefig("reports/correlation_matrix.png", dpi=300, bbox_inches="tight")
plt.show()
```

## 6. Conclusiones Preliminares del EDA

1. **Desbalance de clase**: La tasa de rotación es de ~16%, lo que significa que el dataset está desbalanceado.
2. **Departamentos**: El departamento de Ventas tiene la mayor tasa de rotación.
3. **Horas extras**: Los empleados que hacen horas extras tienen una mayor probabilidad de rotar.
4. **Edad**: Los empleados más jóvenes tienden a rotar más.
5. **Ingreso**: Los empleados con menor ingreso mensual tienen mayor rotación.
6. **Distancia del hogar**: Los empleados que viven más lejos tienden a rotar más.
7. **Años de trabajo**: Los empleados con menos años totales de experiencia tienen mayor rotación.
